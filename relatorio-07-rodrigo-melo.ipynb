### ✍️ Explique (Q1)

O que acontece quando uma chave é repetida no literal — o Python avisa, soma ou sobrescreve? Por que uma lista não pode ser chave, mas uma tupla pode?

**Sua resposta:**

> O Python ao criar um dicionário ele estrutura os dados de forma a nomear uma chave fornecida como imutável e um valor, objeto que pode ser mutável ou não, na memória. Essa chave é criada por método similar ao daquilo que aprendemos em conjuntos: _hash table_. Um cálculo criptográfico adiciona um apontador na memória sobre o valor que armazenamos. Ao manter a chave única, garantimos a unicidade do espaço alocado e faz com que o acesso aos dados seja muito mais rápido.
> Dessa forma temos as seguintes proposições:
> * Ao repetir uma chave no literal com um outro valor e considerando que um dicionário é mutável e a ordem de inserção é preservado a partir do Python 3.7+, simplesmente sobrescreve o valor inicial apontado pela chave. Ela não gera erro, não emite avisos e nem realiza operações com o objeto inicial.
> * Quanto à lista, como é um objeto _unhashable_, isso inviabilizaria a questão de imutabilidade (_hash_) da chave, o que corromperia os dados do dicionário, portanto o interpretador levanta um erro (_TypeError_) sobre essa impossibilidade estrutural. Como a tupla é imutável ela, sim, pode entrar como parâmetros de chaves (_hash_ constante) para um dicionário.

> Quanto à segunda pergunta, a operação de atribuição usando `d[chave] = valor` é feita incondicionalmente, ou seja, se existir a chave, o valor será sobrescrito, caso contrário, será inserido esse novo conjunto `chave:valor`. Já o `d.setdefault(chave, default)` vai fazer a inserção apenas se a chave ainda não existir. Caso contrário, preserva o dado original.
